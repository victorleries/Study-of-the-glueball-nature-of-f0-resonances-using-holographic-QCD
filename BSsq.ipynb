# Numerical solutions for quark-antiquark meson Regge trajectories using Bohr-Sommerfeld semiclassical quantization

First of all on defines the analytical solution using the approximation of considering the masses of the quarks negligible and the potential linear. So the hamiltonian in the CM reference frame is: $H=2p+\sigma r$. And the analytical solution is $M^2=4\sigma\bigl(n+\frac{1}{2}\bigl)$.

```python
def M2_anal(n,sig):
    # n: integer, the radial quantum number. sig: float, the string constant of the gluons.
    return 4*sig*(n+1/2) # returns the mass squared of the quark atiquark meson
    
```

## Bohr-Sommerfeld semiclassical quantization for a more general potential

Now the proposed potential is $V(r)=g\Bigl(\frac{1}{r_0-r}-\frac{1}{r_0}\Bigl)$

```python
def n_M(M,g,r0):
    return 1/(2*np.pi)*((r0-1/(M/g+1/r0))*(M+g/r0)+g*np.log(M*r0/g+1))-1/2
```

```python
n=np.linspace(1,5,5,dtype=int)
r0=np.random.uniform(0,10)
g=np.random.rand()
sigma=g/r0
Mrange=[M2_anal(n[0],sigma),M2_anal(n[len(n)-1],sigma)]
M=np.linspace(np.sqrt(Mrange[0]),np.sqrt(Mrange[1]),10000)

plt.figure()
plt.plot(n,M2_anal(n,sigma),'.',label="Analytical solution for the lineal potential")
plt.plot(n_M(M,g,r0),M**2,label='Numerical solution for more general potential')
plt.xlabel('n')
plt.ylabel('M^2')
plt.grid()
plt.legend()
plt.show()
```
