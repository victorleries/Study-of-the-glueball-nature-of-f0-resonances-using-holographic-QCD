```python
# IMPORT LIBRARIES
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams['text.usetex'] = True
```

# Plot the 3 trajectories of f0 resonances.

```python
# Write data
n=np.arange(1,4)
# Fitted trajectories
trajfit1 = np.array([0.324172, 2.31941, 4.31465])
trajfit2 = np.array([0.876714, 2.84681, 4.8169])
trajfit3 = np.array([1.82009, 3.65477, 5.48945])
# Experimental trajectories + errors
trajexp1 = np.array([0.23, 2.19, 4.38])
trajexp2 = np.array([0.98, 3.06, 4.78])
trajexp3 = np.array([1.81, 3.92, 5.46])
delta1 = np.array([0.13, 0.0096, 0.61])
delta2 = np.array([0.099, 0.2, 0.46])
delta3 = np.array([0.24, 0.46, 0.41])
# Labels of the real masses:
labels1 = [r"$f_0(500)$",r"$f_0(1500)$",r"$f_0(2100)$"]
labels2 = [r"$f_0(980)$",r"$f_0(1710)$",r"$f_0(2200)$"]
labels3 = [r"$f_0(1370)$",r"$f_0(2020)$",r"$f_0(2330)$"]
```

First let's compare experimental trajectories with fitted ones.

```python
plt.figure()
plt.plot(n,trajfit1,label=r"Fitted trajectory 1",c='r')
plt.plot(n,trajfit2,label=r"Fitted trajectory 2",c='g')
plt.plot(n,trajfit3,label=r"Fitted trajectory 3",c='b')
plt.plot(n,trajexp1,'.',c='0')
plt.plot(n,trajexp2,'.',c='0')
plt.plot(n,trajexp3,'.',c='0')
plt.errorbar(n,trajexp1,yerr=delta1,ls='',capsize=3,c='0')
plt.errorbar(n,trajexp2,yerr=delta2,ls='',capsize=3,c='0')
plt.errorbar(n,trajexp3,yerr=delta3,ls='',capsize=3,c='0')
for i in range(len(labels1)):
    plt.text(n[i]+0.05,trajexp1[i],labels1[i])
    plt.text(n[i]+0.05,trajexp2[i],labels2[i])
    plt.text(n[i]+0.05,trajexp3[i],labels3[i])
plt.xlabel(r"$n$")
plt.ylabel(r"$M^2(GeV^2)$")
plt.legend()
plt.grid()
plt.show()
```

Now let's compare the mathematical and physical trajectories.

```python
#The mathematical trajectories are:
trajqq = np.array([1.18941, 2.61941, 4.04941]) # light quarks
trajss = np.array([1.51463, 2.94463, 4.37463]) # strange quarks
trajgg = np.array([0.691112, 3.36291, 6.0347]) # glueballs

# The plot comparing the mathematical with the physical trajectories
plt.figure()
plt.plot(n,trajfit1,label=r"Fitted trajectory 1",c='r')
plt.plot(n,trajfit2,label=r"Fitted trajectory 2",c='g')
plt.plot(n,trajfit3,label=r"Fitted trajectory 3",c='b')
plt.plot(n,trajqq,'--',label=r"$q\overline{q}$ trajectory",c='g')
plt.plot(n,trajss,'--',label=r"$s\overline{s}$ trajectory",c='b')
plt.plot(n,trajgg,'--',label=r"Glueball's trajectory",c='r')
plt.xlabel(r"$n$")
plt.ylabel(r"$M^2(GeV^2)$")
plt.xlim(right=3)
plt.legend()
plt.grid()
plt.show()
```
